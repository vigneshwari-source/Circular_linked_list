# Univarient

# Histogram

```python
plt.figure(figsize=(10,5))
plt.hist(
    df["release_year"],bins=10,color="red",edgecolor="blue",
)
plt.title("Disttibution of release years")
plt.xlabel("release year")
plt.ylabel("count")
plt.show()
```

# Boxplot

```python
plt.figure(figsize=(10,5))
plt.boxplot(df["release_year"])
plt.title("Distribution of release years")
plt.ylabel("count")
plt.show()
```

# Bargraph

```python
plt.figure(figsize=(8,5))
sns.countplot(data=df,
    x='type',
     palette='RdBu'
)
plt.title('count of movies vs TV shows')
plt.xlabel('type')
plt.ylabel('count')
plt.show()
```

# Pie chart

```python
release=df["type"].value_counts()
plt.figure(figsize=(7,7))
plt.pie(
    release,
    labels=release.index,
    autopct="%1.2f%%"
)
plt.title("share type of shows")
plt.show()
```

# Bivarient

# Scatterplot

```python
plt.figure(figsize=(8,6))
plt.scatter(df["release_year"], df["rating"].fillna("Unknown"), color="blue", alpha=0.7)

plt.title("release_year vs rating")
plt.xlabel("release_year")
plt.ylabel("rating")
plt.show()
```

# coorelation

```python
plt.figure(figsize=(10,6))
sns. scatterplot(
    data=df,
    x="release_year",
    y="rating",
    hue="type",
    palette="RdBu"
)
plt.title("release_year vs rating")
plt.xlabel("release_year")
plt.ylabel("rating")
plt.show()
```

# Line plot

```python
plt.figure(figsize=(12, 6))

# Group by release year and content type, then count occurrences
release_trends = df.groupby(['release_year', 'type']).size().unstack().fillna(0)

# Filter to show recent trends (e.g., since 2000) for better visibility
release_trends = release_trends.loc[2000:]

plt.plot(release_trends.index, release_trends['Movie'], marker='o', label='Movies', color='red', linewidth=2)
plt.plot(release_trends.index, release_trends['TV Show'], marker='s', label='TV Shows', color='blue', linewidth=2)

plt.title('Trend of Netflix Content Releases (Year 2000 - 2021)', fontsize=14)
plt.xlabel('Release Year', fontsize=12)
plt.ylabel('Count of Releases', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend()
plt.show()
```

# Reg plot

```python
plt.figure(figsize=(10, 8))
sns.stripplot(
    data=df.dropna(subset=["rating"]),
    x="release_year",
    y="rating",
    alpha=0.5,
    jitter=0.25
)
plt.title("Distribution of Release Years across Ratings")
plt.show()
```
